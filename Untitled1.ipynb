```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample Sales Dataset
data = {
    'Region': ['East', 'West', 'South', 'North', 'East', 'West', 'South', 'North'],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone'],
    'Sales': [50000, 70000, 45000, 65000, 55000, 40000, 75000, 60000],
    'Profit': [8000, 12000, 7000, 10000, 9000, 6000, 13000, 11000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("SALES DATASET")
print(df)

# Calculate KPIs
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

print("\nTotal Sales:", total_sales)
print("Total Profit:", total_profit)

# Region-wise Sales Analysis
region_sales = df.groupby('Region')['Sales'].sum()

print("\nRegion Wise Sales")
print(region_sales)

# Plot Chart
plt.figure(figsize=(8,5))
region_sales.plot(kind='bar')
plt.title("Region Wise Sales Performance")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

print("\nSales Performance Analysis Completed Successfully")
```
